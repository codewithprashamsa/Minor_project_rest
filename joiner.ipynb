```python
import pandas as pd

df = pd.read_csv('data1.csv')

add_sample_hash = {
    1:"Gamming",
    2:"Gamming",
    3:"Web_traffic",
    4:"Web_traffic",
    5:"Web_traffic"
}
modify_id = {
    1:28,
    2:29,
    3:30,
    4:31,
    5:32
}

df["sample_hash"] = df["pcap_id"].map(add_sample_hash)
df['label'] = "bengin"
df["pcap_id"] = df["pcap_id"].map(modify_id)

df.to_csv("data_bengin.csv",index=False)
```
